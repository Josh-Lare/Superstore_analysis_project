```python
# CGPA Calculator
# Author: Joshua Moses Oluwadamilare (Lare)

import pandas as pd
import numpy as np

# --- Student Info ---
name = input("What is your name? ")
age = input("How old are you? ")
school = input("School name: ")
faculty = input("Faculty: ")
dept = input("Department: ")
level = input("Level: ")

# --- Course Data (edit these for any semester) ---
data = {
    "courses": ["Bic_202", "Phy_272", "Nut_202", "Che_276", "Che_259", "Che_226",
                "Ges_201", "Ges_106", "Phy_271", "Psm_201", "Mcb_221", "Bio_216",
                "Bio_215", "Bic_201", "Nut_201", "Ges_102"],
    "unit":   [3, 3, 2, 4, 3, 4, 2, 2, 3, 3, 4, 3, 2, 3, 2, 2],
    "grade_point": [3, 4, 4, 2, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4]
}

df = pd.DataFrame(data)
df["WGP"] = df["unit"] * df["grade_point"]

# --- Remarks ---
conditions = [
    df["grade_point"] == 4,
    df["grade_point"] == 3,
    df["grade_point"] == 2,
    df["grade_point"] == 1
]
choices = ["Excellent", "Good", "Fair", "Poor"]
df["Remarks"] = np.select(conditions, choices, default="F")

# --- CGPA Calculation ---
total_wgp = df["WGP"].sum()
total_unit = df["unit"].sum()
cgpa = round(total_wgp / total_unit, 2)

# --- Class of Degree ---
if cgpa >= 3.5:
    degree_class = "FIRST CLASS"
elif cgpa >= 3.0:
    degree_class = "SECOND CLASS UPPER"
elif cgpa >= 2.0:
    degree_class = "SECOND CLASS LOWER"
elif cgpa >= 1.0:
    degree_class = "THIRD CLASS"
else:
    degree_class = "FAILED"

# --- Output ---
print("=" * 45)
print("        STUDENT ACADEMIC SHEET")
print("=" * 45)
print(f"NAME    : {name}")
print(f"AGE     : {age}")
print(f"SCHOOL  : {school}")
print(f"FACULTY : {faculty}")
print(f"DEPT    : {dept}")
print(f"LEVEL   : {level}")
print("=" * 45)
print(df.to_string(index=False))
print("=" * 45)
print(f"TOTAL WGP   : {total_wgp}")
print(f"TOTAL UNITS : {total_unit}")
print(f"CGPA        : {cgpa}")
print(f"CLASS       : {degree_class}")
print("=" * 45)
```
